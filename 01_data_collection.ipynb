```python
import yfinance as yf
import pandas as pd
import os

companies = {
    "HUL":      "HINDUNILVR.NS",
    "ITC":      "ITC.NS",
    "Nestle":   "NESTLEIND.NS",
    "Britannia":"BRITANNIA.NS",
    "Dabur":    "DABUR.NS",
    "Marico":   "MARICO.NS",
    "Godrej":   "GODREJCP.NS"
}

folder = "C:/Users/Muhsin/Desktop/INTERNSHIP/PROJECTS/gst_fmcg/data/raw/financials"
os.makedirs(folder, exist_ok=True)

for name, ticker in companies.items():
    print(f"Downloading financials for {name}...")
    
    stock = yf.Ticker(ticker)
    
    # Quarterly financials
    df = stock.quarterly_financials
    
    if df.empty:
        print(f"  No data for {name}")
        continue
    
    # Transpose so dates become rows
    df = df.T
    
    # Keep only Total Revenue (may be named differently sometimes)
    if "Total Revenue" in df.columns:
        revenue_df = df[["Total Revenue"]]
    else:
        print(f"  Revenue column not found for {name}")
        continue
    
    revenue_df.to_csv(f"{folder}/{name}_quarterly_revenue.csv")
    
    print(f"  Saved: {len(revenue_df)} rows")

print("\nAll financial data downloaded.")
```
