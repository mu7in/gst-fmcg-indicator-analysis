```python
import yfinance as yf
import pandas as pd
import os

# Define the 7 FMCG companies and their NSE ticker symbols
companies = {
    "HUL":      "HINDUNILVR.NS",
    "ITC":      "ITC.NS",
    "Nestle":   "NESTLEIND.NS",
    "Britannia":"BRITANNIA.NS",
    "Dabur":    "DABUR.NS",
    "Marico":   "MARICO.NS",
    "Godrej":   "GODREJCP.NS"
}

# Date range — GST started July 2017, so we go from there
start_date = "2017-07-01"
end_date   = "2024-12-31"

# Download and save each company's stock data separately
os.makedirs(r"C:\Users\Muhsin\Desktop\INTERNSHIP\PROJECTS\gst_fmcg\data\raw\stock_prices", exist_ok=True)

for name, ticker in companies.items():
    print(rf"Downloading {name} ({ticker})...")
    df = yf.download(ticker, start=start_date, end=end_date)
    df.to_csv(f"C:\Users\Muhsin\Desktop\INTERNSHIP\PROJECTS\gst_fmcg\data\raw\stock_prices{name}_prices.csv")
    print(f"  Saved: {len(df)} rows")

print("\nAll stock data downloaded.")
```
